```python
import numpy as np
import matplotlib.pyplot as plt

print("Setup working!")
```

```python
from math import sqrt

T = 20.0
N = 1000
dt = T / N

rng = np.random.default_rng()
Z = sqrt(dt) * rng.normal(loc=0.0, scale=1.0, size = N)
Z = np.insert(Z, 0, 0)
W = np.cumsum(Z)
```

```python
plt.plot(W)
plt.show()
```

```python
theta = 1.0
sigma = 0.5
mu = 0.0

X = [0] * (N+1)
X[0] = 3.0

for i in range(1, N+1):
    Z = sqrt(dt) * rng.normal(loc=0.0, scale=1.0)
    X[i] = X[i-1] - theta * (X[i-1] - mu) * dt + sigma * Z
```

```python
plt.plot(X)
plt.show()
```

```python
M = 10000
X = np.zeros((M, N + 1))
randlow, randhigh = 2.0, 4.0

X[:, 0] = rng.uniform(randlow, randhigh, size=M)

for i in range(1, N+1):
    Z = sqrt(dt) * rng.normal(loc=0.0, scale=1.0, size=M)
    X[:, i] = X[:, i-1] - theta * (X[:, i-1] - mu) * dt + sigma * Z
```

```python
endvalues = X[:, N]

plt.hist(endvalues, 60, alpha = 0.5, density = True, label = "Simulated Data")
gaussian = rng.normal(loc = mu, scale = sqrt(sigma**2 / (2*theta)), size = M)
plt.hist(gaussian, 60, alpha = 0.5, density = True, label = "Gaussian Distribution")

plt.xlabel("End Value of OU Walk")
plt.ylabel("Density")
plt.legend()
plt.savefig("../figures/stationary_distribution.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
endpoints = X[:, -1]
T = N * dt

theo_mean = mu + (X[:, 0].mean() - mu) * np.exp(-theta * T)
theo_var  = sigma**2 / (2 * theta)
se_mean   = np.sqrt(theo_var / M)

print(f"paths: {M},  dt = {dt},  T = {T}")
print(f"mean:      theory {theo_mean:.5f}   empirical {endpoints.mean():.5f}   (SE {se_mean:.5f})")
print(f"variance:  theory {theo_var:.5f}   empirical {endpoints.var(ddof=1):.5f}")
```
